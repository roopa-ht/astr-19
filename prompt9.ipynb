```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
num = 1000
random = np.random.rand(num)

width = 0.01
min_rand = np.floor(min(random))
max_rand = np.ceil(max(random)) + width

bins = np.arange(min_rand,max_rand, width)

plt.hist(random, bins = bins, alpha = 0.5, edgecolor = 'black')
plt.xlabel('x')
plt.ylabel('y')

plt.savefig("prompt9.pdf", bbox_inches = "tight", dpi = 200)

plt.show()
```
