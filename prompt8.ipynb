```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
steps = 100
x = np.linspace(0,1,steps)
```

```python
def sinx(x):
    sinx = np.sin(x)
    return sinx

def cosx(x):
    cosx = np.cos(x)
    return cosx
```

```python
figure, axes = plt.subplots(1,2)

axes[0].plot(x, sinx(x))
axes[0].set_xlabel('x')
axes[0].set_ylabel('sin(x)')
axes[0].set_title('sin(x) vs. x')

axes[1].plot(x, cosx(x))
axes[1].set_xlabel('x')
axes[1].set_ylabel('cos(x)')
axes[1].set_title('cos(x) vs. x')

figure.subplots_adjust(wspace = 0.4)

plt.savefig("prompt8.pdf", bbox_inches = "tight", dpi = 200)

plt.show()
```
