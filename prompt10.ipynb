```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
num = 1000
shape, scale = 3.0, 4.0
random = np.random.gamma(shape, scale, num)

width = 0.01
min_rand = np.floor(min(random))
max_rand = np.ceil(max(random)) + width

bins = np.arange(min_rand,max_rand, width)

plt.hist(random, bins = bins, alpha = 0.5, linewidth = 2, edgecolor = 'blue')
plt.xlabel('x')
plt.ylabel('y')

plt.savefig("prompt10.pdf", bbox_inches = "tight", dpi = 200)

plt.show()
```
